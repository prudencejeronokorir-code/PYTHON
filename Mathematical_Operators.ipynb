```python
# Assignment operators
marks=100
print("assignment: ", marks)

```

```python
# Comparison operators

# comparing
maths = 90
physics = 75
```

```python
# greater than
print("grater than: ", maths > physics)



```

```python

# greater than or equal to
print("greater than or equal to: ", maths >= physics)

```

```python
# equal to
print("Equal to: ", maths == physics)

```

Logical operators

```python

condition_1 = True
condition_2 = False
```

```python
# And
print("Results of AND OP: ", condition_1 and condition_2)

```

```python
# OR
print("results of OR: ", condition_1 or condition_2)



```

```python
# Not operator
print("results of not: ", not condition_1)

```

```python
# combining and with or
age = 90
is_student = False
```

```python
# combining both
if age >= 90 and not is_student:
    print("You are too old to be a student")
```
